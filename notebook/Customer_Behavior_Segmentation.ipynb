```python
transactions['transaction_date'] = pd.to_datetime(transactions['transaction_date'])
```

```python
transactions['sales_amount'] = transactions['quantity'] * transactions['price']
```

```python
transactions.head()
```

```python
df = transactions.merge(customers, on='customer_id', how='left')
```

```python
df.head()
```

```python
analysis_date = df['transaction_date'].max()
analysis_date
```

```python
rfm = (
    df
    .dropna(subset=['sales_amount'])  # exclude rows with null spend for Monetary
    .groupby('customer_id')
    .agg(
        recency_days = ('transaction_date', lambda x: (analysis_date - x.max()).days),
        frequency     = ('transaction_id', 'nunique'),
        monetary      = ('sales_amount', 'sum')
    )
    .reset_index()
)

rfm.head()
```

```python
## to check if there are any null values or not
rfm[['recency_days','frequency','monetary']].isna().sum()
```

```python
rfm['recency_days'].describe()
```

```python
rfm['R_score'] = pd.qcut(rfm['recency_days'], 4, labels=[4,3,2,1]).astype(int)
rfm['F_score'] = pd.qcut(rfm['frequency'],    4, labels=[1,2,3,4]).astype(int)
rfm['M_score'] = pd.qcut(rfm['monetary'],     4, labels=[1,2,3,4]).astype(int)

rfm['RFM_score'] = rfm['R_score']*100 + rfm['F_score']*10 + rfm['M_score']
```

```python
rfm.head()
```

```python
conditions = [
    (rfm['R_score'] >= 3) & (rfm['F_score'] >= 3) & (rfm['M_score'] >= 3),
    (rfm['R_score'] >= 3) & (rfm['F_score'] <= 2),
    (rfm['R_score'] <= 2) & (rfm['F_score'] >= 3)
]

choices = [
    'High Value / Loyal',
    'New / Promising',
    'At Risk'
]

rfm['segment'] = np.select(conditions, choices, default='Low Engagement')
```

```python
##rfm.drop(columns=['segment'], inplace=True)
```

```python
rfm.head()
```

```python
segment_summary = (
    rfm
    .groupby('segment')
    .agg(
        customers = ('customer_id','count'),
        avg_recency_days = ('recency_days','mean'),
        avg_frequency = ('frequency','mean'),
        total_revenue = ('monetary','sum')
    )
    .reset_index()
)

segment_summary
```

```python
### RFM Methodology
##  Customers were segmented using Recency (days since last purchase event),
##  Frequency (number of purchase events), and Monetary value (total spend).
##  Given the data structure, frequency represents purchase events rather than store visits.
##  RFM scores were computed using quartile-based thresholds to ensure robustness.
```

```python
segment_summary['avg_recency_days'] = segment_summary['avg_recency_days'].round(2)
segment_summary['avg_frequency'] = segment_summary['avg_frequency'].round(2)
segment_summary['total_revenue'] = segment_summary['total_revenue'].round(2)
```
